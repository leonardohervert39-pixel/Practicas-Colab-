```python
celsius = float(input("Ingresa la temperatura en grados Celsius: "))
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius}°C equivalen a {fahrenheit:.2f}°F")
```

Entrada de datos ( celsius = float(...)): Pide al usuario un número, lo lee como texto y lo convierte a decimal ( float).
Conversión ( fahrenheit = ...): Aplica la fórmula matemática (°C × 9/5) + 32 para calcular la equivalencia
Salida de datos ( print(...)): Muestra el resultado formateado, limitando el valor en Fahrenheit a 2 decimales ( :.2f).

```python
kilometros = float(input("Ingresa la distancia en kilómetros: "))
millas = kilometros * 0.621371
print(f"{kilometros} km equivalen a {millas:.2f} millas")
```

Entrada de datos ( kilometros = float(...)): Solicita un valor al usuario y lo transforma a un número decimal ( float).
Conversión ( millas = ...): Multiplica los kilómetros por el factor 0.621371 para obtener la distancia equivalente en millas.
Salida de datos ( print(...)): Imprime el mensaje final en pantalla formateando el resultado en millas a 2 decimales ( :.2f).

```python
total_cuenta = float(input("Ingresa el costo total de la cena ($): "))
porcentaje_propina = float(input("Ingresa el porcentaje de propina (ej. 10, 15, 20): "))
monto_propina = total_cuenta * (porcentaje_propina / 100)
total_a_pagar = total_cuenta + monto_propina
print(f"\nMonto de propina: ${monto_propina:.2f}")
print(f"Total a pagar con propina: ${total_a_pagar:.2f}")
```

Entrada de datos ( total_cuentay porcentaje_propina): Pide al usuario el costo de la cena y el porcentaje que desea dar de propina, convirtiendo ambos valores a números decimales ( float).
Cálculo de propina ( monto_propina = ...): Multiplica el costo total por el porcentaje dividido entre 100 para obtener el dinero exacto de la propina.
Cálculo del total ( total_a_pagar = ...): Suma el costo inicial de la cena más el monto de la propina calculado.
Salida de datos ( print(...)): Imprime en pantalla el dinero de la propina y el total final a pagar, redondeados a 2 decimales ( :.2f).
