# EJERCICIO 1: Determinar cuál es el mayor de 3 números
**1. Definición de variables**

num1, num2, num3: Variables de tipo flotante (float) o entero (int) que almacenarán los tres números ingresados por el usuario.

mayor: Variable que almacenará el valor del número mayor tras realizar las comparaciones.

**2. Descripción en lenguaje natural**

El programa solicita al usuario que ingrese tres números diferentes (o iguales).

Se toma el primer número (num1) y se asume temporalmente como el mayor.

Se compara el segundo número (num2) con el valor actual del mayor. Si num2 es mayor, se actualiza la variable mayor con el valor de num2.

Se compara el tercer número (num3) con la variable mayor. Si num3 es mayor, se actualiza la variable mayor con el valor de num3.

Finalmente, el programa imprime en pantalla el resultado indicando cuál es el número mayor.

(WhatsApp Image 2026-11-03 at 8.05.41 PM.jpeg)

```python
print("--- EJERCICIO 1: Mayor de 3 números ---")
num1 = float(input("Ingresa el primer número: "))
num2 = float(input("Ingresa el segundo número: "))
num3 = float(input("Ingresa el tercer número: "))
mayor = num1
if num2 > mayor:
    mayor = num2
if num3 > mayor:
    mayor = num3
print(f"El número mayor entre {num1}, {num2} y {num3} es: {mayor}")
```

# EJERCICIO 2: Máximo Común Divisor (MCD) de dos números


**1. Definición de variables**

a, b: Variables enteras (int) que representan los dos números ingresados por el usuario para calcular su MCD.

temporal: Variable auxiliar utilizada para almacenar temporalmente un valor durante el intercambio en el bucle del algoritmo.

**2. Descripción en lenguaje natural**

El programa solicita al usuario que ingrese dos números enteros positivos.

Se aplica el Algoritmo de Euclides: se realiza un ciclo mientras el segundo número (b) sea diferente de cero.

Dentro del ciclo, se guarda el valor de b, y el nuevo valor de b pasa a ser el residuo de la división entera de a entre b (a % b), mientras que a toma el valor que tenía b.

El proceso se repite hasta que b sea igual a 0.

Cuando b es 0, el valor almacenado en a es el Máximo Común Divisor, el cual se imprime en pantalla.

WhatsApp Image 2026-11-03 at 8.15.48 PM.jpeg

```python
print("--- EJERCICIO 2: Máximo Común Divisor (MCD) ---")
a = int(input("Ingresa el primer número entero: "))
b = int(input("Ingresa el segundo número entero: "))
num_original_1 = a
num_original_2 = b
while b != 0:
    temporal = b
    b = a % b
    a = temporal
print(f"El Máximo Común Divisor (MCD) de {num_original_1} y {num_original_2} es: {a}")
```


Ingresa el primer número entero: 48

Ingresa el segundo número entero: 18

El Máximo Común Divisor (MCD) de 48 y 18 es: 6
