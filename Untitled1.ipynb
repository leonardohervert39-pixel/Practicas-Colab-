**# 1.-Descripcion Del Problema 1  Precio Final con Descuento**


Este programa solicita al usuario el precio original de un producto y el porcentaje de descuento que se le aplica. Con estos datos,se  calcula el valor que se resta al precio original y muestra en pantalla la cantidad final que debe pagar el cliente.

 # 2.-Pasos
1.Pedir el precio: Le pides al usuario que escriba cuánto cuesta el producto originalmente.

2.Pedir el descuento: Le pides al usuario que escriba el porcentaje de descuento (por ejemplo, 20).

3.Calcular el descuento: Divide el porcentaje entre 100 y lo multiplica por el precio para saber cuántos pesos te van a restaurar.

4.Restar para sacar el precio final: Le restas ese descuento al precio original.

4.Mostrar el resultado: Imprime en pantalla cuánto va a pagar la persona.

```python
precio = float(input("Ingresa el precio del producto: "))
descuento = float(input("Ingresa el porcentaje de descuento: "))
monto_descuento = precio * (descuento / 100)
precio_final = precio - monto_descuento
print("El precio final es:", precio_final)
```
